```python
# Demonstração de Busca Linear em Python puro

def busca_linear(lista, alvo):
  """
  Implementa o algoritmo de busca linear para encontrar um alvo em uma lista.

  Args:
    lista: A lista onde buscar.
    alvo: O elemento a ser encontrado.

  Returns:
    O índice do alvo na lista se encontrado, caso contrário, -1.
  """
  for i in range(len(lista)):
    if lista[i] == alvo:
      return i  # Retorna o índice se o alvo for encontrado
  return -1 # Retorna -1 se o alvo não for encontrado

# Exemplo de uso da busca linear
numeros = [10, 5, 2, 8, 12, 3]
alvo_existente = 8
alvo_nao_existente = 100

indice_existente = busca_linear(numeros, alvo_existente)
indice_nao_existente = busca_linear(numeros, alvo_nao_existente)

print(f"Lista: {numeros}")
print(f"Buscando por {alvo_existente}: Índice {indice_existente}")
print(f"Buscando por {alvo_nao_existente}: Índice {indice_nao_existente}")

# Demonstração de um Sistema Baseado em Regras Simples em Python puro

def classificar_numero(numero):
  """
  Classifica um número como 'pequeno', 'médio' ou 'grande' com base em regras simples.

  Args:
    numero: O número a ser classificado.

  Returns:
    Uma string indicando a classificação do número.
  """
  if numero < 10:
    return "pequeno"
  elif 10 <= numero <= 50:
    return "médio"
  else:
    return "grande"

# Exemplos de uso do sistema baseado em regras
numero1 = 5
numero2 = 30
numero3 = 75

print(f"\nClassificando o número {numero1}: {classificar_numero(numero1)}")
print(f"Classificando o número {numero2}: {classificar_numero(numero2)}")
print(f"Classificando o número {numero3}: {classificar_numero(numero3)}")
```

# Introdução ao LangChain
LangChain é um framework poderoso projetado para simplificar o desenvolvimento de aplicações que utilizam Large Language Models (LLMs). Ele permite que os desenvolvedores conectem LLMs com outras fontes de dados e interajam com o ambiente, tornando a criação de aplicações mais complexas e contextuais muito mais fácil.

Propósito e Recursos Principais:

* Encadeamento de Componentes (Chains): LangChain permite encadear diferentes componentes, como modelos de linguagem, prompts, ferramentas e memória, para criar fluxos de trabalho complexos. Isso facilita a construção de aplicações que exigem múltiplas etapas de processamento.
* Suporte a Vários LLMs: O framework oferece interfaces unificadas para interagir com diversos modelos de linguagem, independentemente do provedor (OpenAI, Hugging Face, etc.), permitindo flexibilidade na escolha do modelo.
Gerenciamento de Prompts: LangChain fornece ferramentas para criar, gerenciar e otimizar prompts, o que é crucial para obter o desempenho desejado dos LLMs.
* Memória: O framework inclui mecanismos para adicionar memória aos aplicativos, permitindo que os modelos "lembrem" de interações anteriores em uma conversa.
* Agentes e Ferramentas: LangChain facilita a criação de agentes que podem decidir qual ação tomar com base na entrada do usuário, utilizando um conjunto de ferramentas (como busca na web, calculadoras, etc.) para interagir com o mundo exterior.
* Ao fornecer uma interface padronizada e um rico conjunto de componentes pré-construídos, LangChain acelera significativamente o desenvolvimento de aplicações baseadas em LLMs, permitindo a criação de soluções mais robustas e inovadoras.

```python
import os
# Certifique-se de ter a chave da API da OpenAI configurada como uma variável de ambiente
# Ex: export OPENAI_API_KEY='sua-chave-aqui'
# Ou descomente e substitua pela sua chave (não recomendado para produção)
# os.environ["OPENAI_API_KEY"] = "SUA_CHAVE_AQUI"

from langchain_openai import ChatOpenAI
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain

# 2. Instanciar um modelo de linguagem
# Substitua pelo modelo de sua preferência e certifique-se de ter as dependências instaladas
# Ex: pip install langchain-openai
try:
    llm = ChatOpenAI(temperature=0.7) # Usando um modelo de chat da OpenAI
    print("Modelo de linguagem instanciado com sucesso.")
except Exception as e:
    print(f"Erro ao instanciar o modelo de linguagem: {e}")
    llm = None


# 3. Demonstrar interação básica com o modelo
if llm:
    try:
        print("\nDemonstração de interação básica com o LLM:")
        prompt_texto = "Qual a capital da França?"
        response = llm.invoke(prompt_texto) # Use invoke para interação simples
        print(f"Prompt: {prompt_texto}")
        print(f"Resposta: {response.content}") # A resposta de modelos de chat geralmente tem um atributo .content
    except Exception as e:
        print(f"Erro durante a interação básica com o LLM: {e}")

# 4. Criar um template de prompt
prompt_template_texto = "Escreva uma frase curta sobre {topico}."
prompt = PromptTemplate(
    input_variables=["topico"],
    template=prompt_template_texto,
)
print(f"\nTemplate de prompt criado:\n{prompt_template_texto}")

# 5. Instanciar uma Chain simples (LLMChain)
if llm:
    try:
        chain = LLMChain(llm=llm, prompt=prompt)
        print("\nChain (LLMChain) instanciada com sucesso.")

        # 6. Executar a chain com um valor de entrada
        topico_exemplo = "inteligência artificial"
        print(f"\nExecutando a chain com o tópico: {topico_exemplo}")
        response_chain = chain.invoke({"topico": topico_exemplo}) # Use invoke para chains
        print(f"Resposta da chain: {response_chain['text']}") # O output da LLMChain geralmente está em ['text']
    except Exception as e:
        print(f"Erro ao executar a chain: {e}")
else:
    print("\nChain não pode ser instanciada porque o LLM não foi carregado.")
```
